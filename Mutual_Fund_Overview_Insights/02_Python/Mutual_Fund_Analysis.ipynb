#STAGE 3 – Data Normalization

```python
#Apply MinMaxScaler
scaler = MinMaxScaler()

scale_cols = [
    'returns_3yr',
    'expense_ratio',
    'sd'
]

ranking_df[[
    'returns_scaled',
    'expense_scaled',
    'sd_scaled'
]] = scaler.fit_transform(
    ranking_df[scale_cols].assign(
        sd=ranking_df['sd'].fillna(
            ranking_df['sd'].median()
        )
    )
)
```

```python
#Reverse the expense and risk scores
ranking_df['expense_score'] = (
    1 - ranking_df['expense_scaled']
)

ranking_df['risk_score'] = (
    1 - ranking_df['sd_scaled']
)
```

```python
#Check the normalized values
ranking_df[[
    'returns_3yr',
    'returns_scaled',
    'expense_ratio',
    'expense_score',
    'sd',
    'risk_score'
]].head(10)
```

#STAGE 4 – Fund Scoring & Ranking

```python
#Create the age suitability score
ranking_df['age_score'] = (
    1 - abs(ranking_df['fund_age_yr'] - 7.5) / 7.5
).clip(0, 1)
```

```python
#Create the consistency score
ranking_df['consistency_score'] = np.where(
    ranking_df['returns_1yr'] > 0,
    1,
    0
)
```

```python
#Calculate the final score
ranking_df['final_score'] = (

    ranking_df['returns_scaled'] * 0.45 +

    ranking_df['expense_score'] * 0.20 +

    ranking_df['age_score'] * 0.15 +

    ranking_df['consistency_score'] * 0.10 +

    ranking_df['risk_score'] * 0.10

)
```

```python
#Rank all eligible funds
ranking_df = ranking_df.sort_values(
    by='final_score',
    ascending=False
)

ranking_df['rank'] = range(
    1,
    len(ranking_df) + 1
)
```

```python
#Display the ranking
ranking_df[[
    'rank',
    'scheme_name',
    'category',
    'returns_1yr',
    'returns_3yr',
    'expense_ratio',
    'fund_age_yr',
    'sd',
    'final_score'
]].head(30)
```

```python
#Understand which factors drive high scores
ranking_df[[
    'returns_scaled',
    'expense_score',
    'age_score',
    'consistency_score',
    'risk_score'
]].corr()
```
